```python
import mysql.connector

conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="Avdya@7468",   
    database="student_management"
)

cursor = conn.cursor()
print("Connected to database")
```

```python
query = """
SELECT s.student_id, s.name, c.course_name, s.marks, s.grade
FROM students s
JOIN courses c
ON s.course_id = c.course_id;
"""

cursor.execute(query)

rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
import pandas as pd

df = pd.read_sql(query, conn)
df
```

```python
insert_query = """
INSERT INTO students (student_id, name, marks, grade, course_id)
VALUES (%s, %s, %s, %s, %s)
"""

data = (6, "Ananya Deshmukh", 88, "A", 101)

cursor.execute(insert_query, data)
conn.commit()

print("New student added")
```

```python
update_query = """
UPDATE students
SET marks = %s, grade = %s
WHERE student_id = %s
"""

cursor.execute(update_query, (75, "B", 2))
conn.commit()

print("Marks updated")
```
